Q1. Generate 3D wireframe plots for the following functions:

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
```

```python
# f(x, y) = e^(x² + y²) for x, y ∈ [0, 2π]

x = np.linspace(0,2*np.pi,50)
y = np.linspace(0,2*np.pi,50)
X,Y = np.meshgrid(x,y)
Z = np.exp(X**2 + Y**2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X,Y,Z,color='black')
plt.show()
```

```python
# f(x, y) = e^(x² − y²) for x, y ∈ [0, 2]

x = np.linspace(0,2,50)
y = np.linspace(0,2,50)
X,Y = np.meshgrid(x,y)
Z = np.exp(X**2 - Y**2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X,Y,Z,color='cyan',linewidth=0.8)
plt.show()
```

```python
# f(x, y) = e^(−x² − y²) for x, y ∈ [0, 2π]

x = np.linspace(0,2*np.pi,50)
y = np.linspace(0,2*np.pi,50)
X,Y = np.meshgrid(x,y)
Z = np.exp(-X**2 - Y**2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X,Y,Z,color='purple')
plt.show()
```

```python
# f(x, y) = x*e^(−x² − y²) for x, y ∈ [−6, 6]

x = np.linspace(-6,6,50)
y = np.linspace(-6,6,50)
X,Y = np.meshgrid(x,y)
Z = X*np.exp(-X**2 - Y**2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X,Y,Z,color='blue', linewidth=1.5, alpha=0.7)
plt.show()
```

```python
# f(x, y) = sin(X) + cos(Y) for x, y ∈ [−5, 5]

x = np.linspace(-5,5,50)
y = np.linspace(-5,5,50)
X,Y = np.meshgrid(x,y)
Z = np.sin(X) + np.cos(Y)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X,Y,Z,color='red', linewidth=1.5, alpha=0.7)
plt.show()
```

Q2. Generate 3D surface plots for the following functions:

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
```

```python
# f(x, y) = x*e^(y) for x, y ∈ [0, 2]

x = np.linspace(0,2,50)
y = np.linspace(0,2,50)
X,Y = np.meshgrid(x,y)
Z = X*np.exp(Y)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X,Y,Z,cmap='viridis')
plt.show()
```

```python
# f(x, y) = x³+y³+x²+y²+x+y+5 for x, y ∈ [−2, 2]

x = np.linspace(-2,2,50)
y = np.linspace(-2,2,50)
X,Y = np.meshgrid(x,y)
Z = X**3+Y**3+X**2+Y**2+X+Y+5

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X,Y,Z,cmap='plasma')
plt.show()
```

```python
# f(x, y) = x²+y² for x, y ∈ [−6, 6]

x = np.linspace(-6,6,50)
y = np.linspace(-6,6,50)
X,Y = np.meshgrid(x,y)
Z = X**2 + Y**2

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X,Y,Z,cmap='inferno')
plt.show()
```

```python
# f(x, y) = cos(|x| + |y|) for x, y ∈ [−1, 1]

x = np.linspace(-1,1,50)
y = np.linspace(-1,1,50)
X,Y = np.meshgrid(x,y)
Z = np.cos(np.abs(X)+np.abs(Y))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X,Y,Z,cmap='coolwarm')
plt.show()
```

```python
# f(x, y) = cos(X² + Y² - 0.5) - 0.5 for x, y ∈ [−1, 1]

x = np.linspace(-1,1,50)
y = np.linspace(-1,1,50)
X,Y = np.meshgrid(x,y)
Z = np.cos(X**2 + Y**2 - 0.5) - 0.5

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X,Y,Z,cmap='jet')
plt.show()
```
