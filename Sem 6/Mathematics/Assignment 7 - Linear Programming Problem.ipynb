```python
import pulp
from pulp import *
```

```python
# (a) Max z = 150x + 75y
# Subject to: 4x + 6y ≤ 24
#             5x + 3y ≤ 15
#             x ≥ 0, y ≥ 0

model = LpProblem("Q_a", LpMaximize)

x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)

model += 150*x + 75*y
model += 4*x + 6*y <= 24
model += 5*x + 3*y <= 15

model.solve()
print("x =", x.value(), "y =", y.value(), "Max z =", value(model.objective))
```

```python
# (b) Min z = 3.5x + 2y
# Subject to: x + y ≥ 5
#             x ≥ 4
#             y ≤ 2
#             x, y ≥ 0

model = LpProblem("Q_b", LpMinimize)

x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)

model += 3.5*x + 2*y
model += x + y >= 5
model += x >= 4
model += y <= 2

model.solve()
print("x =", x.value(), "y =", y.value(), "Min z =", value(model.objective))
```

```python
# (c) Max z = 3x + 5y + 4z
# Subject to: 2x + 3y ≤ 8
#             2y + 5z ≤ 10
#             3x + 2y + 4z ≤ 15
#             x, y, z ≥ 0

model = LpProblem("Q_c", LpMaximize)

x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)
z = LpVariable("z", lowBound=0)

model += 3*x + 5*y + 4*z
model += 2*x + 3*y <= 8
model += 2*y + 5*z <= 10
model += 3*x + 2*y + 4*z <= 15

model.solve()
print("x =", x.value(), "y =", y.value(), "z =", z.value(), "Max z =", value(model.objective))
```

```python
# (d) Min z = x + 2y + z
# Subject to: x + 1/2 y + 1/2 z ≤ 1
#             3/2 x + 2y + z ≥ 8
#             x, y, z ≥ 0

model = LpProblem("Q_d", LpMinimize)

x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)
z = LpVariable("z", lowBound=0)

model += x + 2*y + z
model += x + 0.5*y + 0.5*z <= 1
model += 1.5*x + 2*y + z >= 8

model.solve()
print("x =", x.value(), "y =", y.value(), "z =", z.value(), "Min z =", value(model.objective))
```

```python
# (e) Min z = x + y
# Subject to: x ≥ 6
#             y ≥ 6
#             x + y ≤ 11
#             x, y ≥ 0

model = LpProblem("Q_e", LpMinimize)

x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)

model += x + y
model += x >= 6
model += y >= 6
model += x + y <= 11

model.solve()
print("x =", x.value(), "y =", y.value(), "Min z =", value(model.objective))
```

```python
# (f) Max z = x + y
# Subject to: x − y ≥ 1
#             x + y ≥ 2
#             x, y ≥ 0

model = LpProblem("Q_f", LpMaximize)

x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)

model += x + y
model += x - y >= 1
model += x + y >= 2

model.solve()
print("x =", x.value(), "y =", y.value(), "Max z =", value(model.objective))
```

```python
# (g) Max z = 4x + y + 3z + 5w
# Subject to: 4x + 6y − 5z − 4w ≥ −20
#            −3x − 2y + 4z + w ≤ 10
#            −8x − 3y + 3z + 2w ≤ 20
#            x, y, z, w ≥ 0

model = LpProblem("Q_g", LpMaximize)

x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)
z = LpVariable("z", lowBound=0)
w = LpVariable("w", lowBound=0)

model += 4*x + y + 3*z + 5*w
model += 4*x + 6*y - 5*z - 4*w >= -20
model += -3*x - 2*y + 4*z + w <= 10
model += -8*x - 3*y + 3*z + 2*w <= 20

model.solve()
print("x =", x.value(), "y =", y.value(), "z =", z.value(), "w =", w.value(), "Max z =", value(model.objective))
```
