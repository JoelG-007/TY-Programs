# Assignment 7: Linear Programming Problem

#### Q(a) - Max z = 150x + 75y

```python
from pulp import *

model = LpProblem("a", LpMaximize)
x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)

model += 150*x + 75*y
model += 4*x + 6*y <= 24
model += 5*x + 3*y <= 15

print(model)
status = value(model.solve())
if status == -1:
    print("Infeasible")
elif status == -2:
    print("Unbounded")
else:
    print("Solution Status =", model.solve())
    print("\nx =", x.value(), ", y =", y.value())
    print("Max z =", value(model.objective))
```
