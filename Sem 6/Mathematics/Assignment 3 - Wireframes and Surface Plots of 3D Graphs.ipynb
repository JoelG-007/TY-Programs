# Assignment 3: Wireframes and Surface Plots of 3D Graphs

#### Q1 - Wireframe Plots

##### Q1.1 - f(x,y) = e^(x²+y²),  x,y ∈ [0, 2π]

```python
import numpy as np
import matplotlib.pyplot as plt

x = y = np.linspace(0, 2*np.pi, 50)
X, Y = np.meshgrid(x, y)
Z = np.exp(X**2 + Y**2)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X, Y, Z, linewidth=0.7)
ax.set(xlabel='X axis', ylabel='Y axis', zlabel='Z axis', title='f(x,y) = e^(x²+y²)')
plt.show()
```

#### Q2 - Surface Plots

##### Q2.1 - f(x,y) = x·eʸ,  x,y ∈ [0, 2]

```python
import numpy as np
import matplotlib.pyplot as plt

x = y = np.linspace(0, 2, 50)
X, Y = np.meshgrid(x, y)
Z = X * np.exp(Y)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set(xlabel='X axis', ylabel='Y axis', zlabel='Z axis', title='f(x,y) = x·eʸ')
plt.show()
```
