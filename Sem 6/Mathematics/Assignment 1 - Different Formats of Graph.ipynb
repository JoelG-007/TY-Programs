#### Q1 - Contour plot of z = x² + y²

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

plt.contourf(X, Y, Z)
plt.colorbar()
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("z = x² + y²")
plt.show()
```

#### Q2 - Contour plot of z = sin(x) + cos(y)

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 100)
y = np.linspace(-2*np.pi, 2*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) + np.cos(Y)

plt.contourf(X, Y, Z)
plt.colorbar()
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("z = sin(x) + cos(y)")
plt.show()
```

#### Q3 - Line graph: Student marks in 5 subjects

```python
import matplotlib.pyplot as plt

subjects = ["Maths", "CS", "Physics", "Chemistry", "English"]
marks = [78, 88, 72, 69, 85]

plt.plot(subjects, marks, marker='o')
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Marks in 5 Subjects")
plt.show()
```

#### Q4 - Two lines: Sales of Product A & B over 6 months

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
prodA = [50, 60, 55, 70, 65, 80]
prodB = [30, 40, 45, 50, 60, 75]

plt.plot(months, prodA, marker='o', label="Product A")
plt.plot(months, prodB, marker='o', label="Product B")
plt.xlabel("Months")
plt.ylabel("Product Sales")
plt.title("Sales of Product A & B")
plt.legend()
plt.show()
```

#### Q5 - Bar graph: Number of students in 5 classes

```python
import matplotlib.pyplot as plt

cls = ["FY", "SY", "TY", "M.Sc.-I", "M.Sc.-II"]
students = [120, 110, 95, 80, 70]

plt.bar(cls, students)
plt.xlabel("Class")
plt.ylabel("Number of Students")
plt.title("Number of Students in Classes")
plt.show()
```

#### Q6 - Bar graph: Average rainfall in 5 months

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
rainfall = [80, 60, 100, 150, 120]

plt.bar(months, rainfall)
plt.xlabel("Months")
plt.ylabel("Rainfall (mm)")
plt.title("Average Rainfall in 5 Months")
plt.show()
```

#### Q7 - Histogram: Marks obtained by students

```python
import matplotlib.pyplot as plt

marks = [5]*3 + [15]*5 + [25]*9 + [35]*14 + [45]*11 + [55]*8

plt.hist(marks, bins=6, edgecolor='black')
plt.xlabel("Marks Interval")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

#### Q8 - Histogram: Distribution of weights

```python
import matplotlib.pyplot as plt

weights = [42]*5 + [47]*9 + [52]*14 + [57]*10 + [62]*7

plt.hist(weights, bins=[40, 45, 50, 55, 60, 65], edgecolor='black')
plt.xlabel("Weight (kg)")
plt.ylabel("Number of Persons")
plt.title("Distribution of Weights")
plt.show()
```

#### Q9 - Pie chart: % of population by grain

```python
import matplotlib.pyplot as plt

grain = ["Wheat", "Bajra", "Rice", "Maize", "Jowar", "Others"]
per = [23, 25, 22, 15, 11, 4]

plt.pie(per, labels=grain, autopct="%1.1f%%")
plt.title("% of Population by Grain")
plt.show()
```

#### Q10 - Pie chart: Daily activity time

```python
import matplotlib.pyplot as plt

activity = ["Sleep", "Study", "Mobile", "Entertainment", "Others"]
hours = [8, 4, 3, 2, 7]

plt.pie(hours, labels=activity, autopct="%1.1f%%")
plt.title("Daily Time Distribution")
plt.show()
```
