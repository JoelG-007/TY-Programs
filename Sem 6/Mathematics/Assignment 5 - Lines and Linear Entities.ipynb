# Assignment 5: Lines and Linear Entities

```python
from sympy import *
from sympy.geometry import *
x, y = symbols('x y')
```

#### Q1 - Rotate line through (0,0) and (0,1) by 30°, find equation

```python
L = Line(Point(0, 0), Point(0, 1))
Lr = L.rotate(pi/6)
print("Rotated Line:", Lr)
print("Equation:", Lr.equation())
```

#### Q2 - Rotate segment (1,0)–(2,-1) by 180°

```python
S = Segment(Point(1, 0), Point(2, -1))
print("Rotated Segment:", S.rotate(pi))
```

#### Q3 - Rotate ray from (0,0) toward (4,4) by 90° clockwise & anticlockwise

```python
R = Ray(Point(0, 0), Point(4, 4))
print("Anticlockwise 90°:", R.rotate(pi/2))
print("Clockwise 90°:", R.rotate(-pi/2))
```

#### Q4 - Reflect line 4x+3y=5 through line x+y=0, find equation

```python
L = Line(4*x + 3*y - 5)
Lr = L.reflect(Line(x + y))
print("Reflected Line:", Lr)
print("Equation:", Lr.equation())
```

#### Q5 - Reflect segment (2,3)–(4,6) through line 7x+6y=3

```python
S = Segment(Point(2, 3), Point(4, 6))
print("Reflected Segment:", S.reflect(Line(7*x + 6*y - 3)))
```

#### Q6 - Reflect ray from (0,0) toward (2,4) through line x-2y=3

```python
R = Ray(Point(0, 0), Point(2, 4))
print("Reflected Ray:", R.reflect(Line(x - 2*y - 3)))
```

#### Q7 - Line A[2,1], B[4,-1] transformed by T=[[1,2],[2,1]], find equation

```python
T = Matrix([[1, 2], [2, 1]])
A = Point(T * Matrix([2, 1]))
B = Point(T * Matrix([4, -1]))
print("Transformed Line equation:", Line(A, B).equation())
```

#### Q8 - Segment A[1,1], B[-4,-1] transformed by T=[[1,-2],[-2,1]], find endpoints

```python
T = Matrix([[1, -2], [-2, 1]])
A = Point(T * Matrix([1, 1]))
B = Point(T * Matrix([-4, -1]))
print("A* =", A)
print("B* =", B)
```

#### Q9 - Segment A[2,5], B[4,-13] transformed by T=[[2,3],[4,1]], find midpoint & slope

```python
T = Matrix([[2, 3], [4, 1]])
A = Point(T * Matrix([2, 5]))
B = Point(T * Matrix([4, -13]))
S = Segment(A, B)
print("Midpoint:", S.midpoint)
print("Slope:", S.slope)
```

#### Q10 - Shear line 2x+y=3 in x & y by 2 & -3, find equation

```python
T = Matrix([[1, 2], [-3, 1]])
P1 = Point(0, 3)
P2 = Point(1, 1)
P1t = Point(T * Matrix(list(P1)))
P2t = Point(T * Matrix(list(P2)))
print("Equation of resulting line:", Line(P1t, P2t).equation())
```
