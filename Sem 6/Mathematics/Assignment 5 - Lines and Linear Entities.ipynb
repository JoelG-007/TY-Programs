```python
from sympy import *
from sympy.geometry import *
x, y = symbols('x y')
```

```python
# Q.1) Rotate the line by 30 degrees having two points (0, 0) and (0, 1). Also find its equation after applying rotation.
L = Line(Point(0,0), Point(0,1))
Lr = L.rotate(pi/6)
print(Lr)
print(Lr.equation())
```

```python
# Q.2) Rotate the segment by 180 degrees having end points (1, 0) and (2, −1).
S = Segment(Point(1,0), Point(2,-1))
print(S.rotate(pi))
```

```python
# Q.3) Rotate the ray by 90 degrees clockwise and anticlockwise having starting point (0, 0) in the direction of (4, 4).
R = Ray(Point(0,0), Point(4,4))
print("Anticlockwise:", R.rotate(pi/2))
print("Clockwise:", R.rotate(-pi/2))
```

```python
# Q.4) Reflect the line 4x + 3y = 5 through the line x + y = 0 and find the equation of the reflected line.
L = Line(4*x + 3*y - 5)
print(L.reflect(Line(x + y)))
```

```python
# Q.5) Reflect the segment having two endpoints (2, 3) and (4, 6) through the line 7x + 6y = 3.
S = Segment(Point(2,3), Point(4,6))
print(S.reflect(Line(7*x + 6*y - 3)))
```

```python
# Q.6) Reflect the ray having starting point (0, 0) in the direction of (2, 4) through the line x − 2y = 3.
R = Ray(Point(0,0), Point(2,4))
print(R.reflect(Line(x - 2*y - 3)))
```

```python
'''
Q.7) If the line with points A [2, 1], B [4, -1] is transformed by the transformation matrix
[T] = [ 1 2 ]
      [ 2 1 ]
then find the equation of the transformed line.
'''
T = Matrix([[1,2],[2,1]])
A = Point(T*Matrix([2,1]))
B = Point(T*Matrix([4,-1]))
print(Line(A,B).equation())
```

```python
'''
Q.8) If the line segment with endpoints A [1, 1], B [-4, -1] is transformed by the transformation matrix
[T] = [ 1 -2 ]
      [ -2 1 ]
then find the end points of the transformed line.
'''
T = Matrix([[1,-2],[-2,1]])
A = Point(T*Matrix([1,1]))
B = Point(T*Matrix([-4,-1]))
print(A, B)
```

```python
'''
Q.9) If the line segment joining the points A [2, 5], B [4, -13] is transformed to the line segment AB by the transformation matrix
[T] = [ 2 3 ]
      [ 4 1 ]
then find
- the midpoint of AB
- the slope of AB.
'''
T = Matrix([[2,3],[4,1]])
A = Point(T*Matrix([2,5]))
B = Point(T*Matrix([4,-13]))
S = Segment(A,B)
print("Midpoint:", S.midpoint)
print("Slope:", S.slope)
```

```python
'''
Q.10) If we apply shearing on the line 2x + y = 3 in x and y direction by 2 and -3 units respectively, 
then find the equation of the resulting line.
'''
T = Matrix([[1,2],[-3,1]])

P1 = Point(0,3)
P2 = Point(1,1)

P1t = Point(T*Matrix(P1))
P2t = Point(T*Matrix(P2))

print(Line(P1t,P2t).equation())
```
