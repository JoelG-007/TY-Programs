# Assignment 6: Polygon and Its Important Attributes

```python
from sympy import *
x, y = symbols('x y')
```

#### Q1 - Polygon (0,0),(1,0),(2,2),(1,4) - area & perimeter

```python
P = Polygon(Point(0,0), Point(1,0), Point(2,2), Point(1,4))
print("Area:", P.area)
print("Perimeter:", P.perimeter)
```

#### Q2 - Regular polygon: 8 sides, radius 5, centered at origin

```python
P = RegularPolygon(Point(0, 0), 5, 8)
print("Area:", P.area)
print("Perimeter:", P.perimeter)
```

#### Q3 - Regular polygon: 7 sides, radius 1.5, centered (2,2), reflect through x-y=5

```python
P = RegularPolygon(Point(2, 2), 1.5, 7)
print("Reflected Polygon:", P.reflect(Line(x - y - 5)))
```

#### Q4 - Polygon (0,0),(2,0),(2,3),(1,6) rotated by 180°

```python
P = Polygon(Point(0,0), Point(2,0), Point(2,3), Point(1,6))
print("Rotated Polygon:", P.rotate(pi))
```

#### Q5 - Reflect triangle A[1,0], B[2,-1], C[-1,3] through line y=3

```python
T = Triangle(Point(1, 0), Point(2, -1), Point(-1, 3))
print("Reflected Triangle:", T.reflect(Line(y - 3)))
```

#### Q6 - Angles at each vertex of triangle A[0,0], B[2,2], C[0,2]

```python
T = Triangle(Point(0, 0), Point(2, 2), Point(0, 2))
print("Angles:", T.angles)
```

#### Q7 - Check isosceles, right-angled, scalene

##### Q7.i - A[0,0], B[4,0], C[2,4]

```python
T = Triangle(Point(0,0), Point(4,0), Point(2,4))
print("Isosceles:", T.is_isosceles(), "| Right:", T.is_right(), "| Scalene:", T.is_scalene())
```

##### Q7.ii - A[0,0], B[4,0], C[4,3]

```python
T = Triangle(Point(0,0), Point(4,0), Point(4,3))
print("Isosceles:", T.is_isosceles(), "| Right:", T.is_right(), "| Scalene:", T.is_scalene())
```

##### Q7.iii - A[0,0], B[4,0], C[1,4]

```python
T = Triangle(Point(0,0), Point(4,0), Point(1,4))
print("Isosceles:", T.is_isosceles(), "| Right:", T.is_right(), "| Scalene:", T.is_scalene())
```
