```python
import pulp
from pulp import LpProblem, LpVariable, LpInteger, lpSum, LpMinimize, value
```

```python
# Reference Question:
# A boutique brewery has two warehouses A and B with supplies 1000 and 4000 cases.
# Bars 1–5 demand 500, 900, 1800, 200 and 700 cases respectively.
# Minimize transportation cost.

Warehouses = ["A", "B"]
Bars = ["1", "2", "3", "4", "5"]

Supply = {"A": 1000, "B": 4000}
Demand = {"1": 500, "2": 900, "3": 1800, "4": 200, "5": 700}

Costs = {
    "A": {"1": 2, "2": 4, "3": 5, "4": 2, "5": 1},
    "B": {"1": 3, "2": 1, "3": 3, "4": 2, "5": 3}
}

prob = LpProblem("Beer_Distribution", LpMinimize)

x = LpVariable.dicts("Route", (Warehouses, Bars), 0, None, LpInteger)

prob += lpSum(x[w][b] * Costs[w][b] for w in Warehouses for b in Bars)

for w in Warehouses:
    prob += lpSum(x[w][b] for b in Bars) <= Supply[w]

for b in Bars:
    prob += lpSum(x[w][b] for w in Warehouses) >= Demand[b]

prob.solve()

print("Minimum Cost =", value(prob.objective))
for v in prob.variables():
    print(v.name, "=", v.varValue)
```

```python
# SET 1:
# The ICARE company has three plants A, B, C with supplies 50, 75, 25 gallons.
# Retail shops R1, R2, R3, R4 demand 20, 20, 50, 60 gallons respectively.
# Minimize transportation cost.

Plants = ["A", "B", "C"]
Shops = ["R1", "R2", "R3", "R4"]

Supply = {"A": 50, "B": 75, "C": 25}
Demand = {"R1": 20, "R2": 20, "R3": 50, "R4": 60}

Costs = {
    "A": {"R1": 3, "R2": 5, "R3": 7, "R4": 6},
    "B": {"R1": 2, "R2": 5, "R3": 8, "R4": 2},
    "C": {"R1": 3, "R2": 6, "R3": 9, "R4": 2}
}

prob = LpProblem("ICARE_Transport", LpMinimize)

x = LpVariable.dicts("Route", (Plants, Shops), 0)

prob += lpSum(x[p][s] * Costs[p][s] for p in Plants for s in Shops)

for p in Plants:
    prob += lpSum(x[p][s] for s in Shops) <= Supply[p]

for s in Shops:
    prob += lpSum(x[p][s] for p in Plants) >= Demand[s]

prob.solve()

print("Minimum Cost =", value(prob.objective))
for v in prob.variables():
    print(v.name, "=", v.varValue)
```
