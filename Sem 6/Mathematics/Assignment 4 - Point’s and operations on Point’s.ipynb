Q.1) Check whether given three points are co-linear or not.

```python
# i) A [2, 2, 2], B [-1, -1, -1], C [3, 4, -7]

A=Point(2,2,2); B=Point(-1,-1,-1); C=Point(3,4,-7)
print(Point.is_collinear(A,B,C))
```

```python
# ii) A [0, 0, 0], B [-2, -2, -2], C [-7, -7, -7]
A=Point(0,0,0); B=Point(-2,-2,-2); C=Point(-7,-7,-7)
print(Point.is_collinear(A,B,C))
```

Q.2) Check whether given three points are co-planar or not.

```python
# i) A [2, 2, 2], B [−1, −1, −1], C [3, 4, −7]
A=Point(2,2,2); B=Point(-1,-1,-1); C=Point(3,4,-7)
print(Point.are_coplanar(A,B,C))
```

```python
# ii) A [0, 0, 0], B [−2, −2, −2], C [−7, −7, −7]
A=Point(0,0,0); B=Point(-2,-2,-2); C=Point(-7,-7,-7)
print(Point.are_coplanar(A,B,C))
```

Q.3) Find the distance between points x,y ; y,w and x,z if x = [0, 0], y = [2, 2], z = [-1, -1], w = [3, 4]

```python
x=Point(0,0); y=Point(2,2); z=Point(-1,-1); w=Point(3,4)

print(x.distance(y))
print(y.distance(w))
print(x.distance(z))
```

Q.4) Apply each of the following transformations on the point P [-2, 3, 5].

```python
# Method 1
P=Point(-2,3,5)

# a) Reflection Y-axis
print(Point(-P.x,P.y,P.z))

# b) Scale X by 2
print(Point(2*P.x,P.y,P.z))

# c) Scale Y by 1.5
print(Point(P.x,1.5*P.y,P.z))

# d) Shear Y by 2
print(Point(P.x,P.y+2*P.x,P.z))

# e) Scale X by 1/2, Y by 3
print(Point(P.x/2,3*P.y,P.z))

# f) Shear X=-2, Y=4
print(Point(P.x-2*P.y,P.y+4*P.x,P.z))

# g) Rotation 30°
theta=pi/6
print(Point(P.x*cos(theta)-P.y*sin(theta),P.x*sin(theta)+P.y*cos(theta),P.z))
```

```python
# Method 2 
P = Point(-2,3,5)

# a) Reflection through Y-axis
print(P.reflect(Line(Point(0,0),Point(0,1))))

# b) Scaling in X by 2
print(Point(Matrix([[2,0,0],[0,1,0],[0,0,1]])*Matrix(P)))

# c) Scaling in Y by 1.5
print(Point(Matrix([[1,0,0],[0,1.5,0],[0,0,1]])*Matrix(P)))

# d) Shearing in Y by 2
print(Point(Matrix([[1,0,0],[2,1,0],[0,0,1]])*Matrix(P)))

# e) Scaling X by 1/2 and Y by 3
print(Point(Matrix([[1/2,0,0],[0,3,0],[0,0,1]])*Matrix(P)))

# f) Shearing X by −2 and Y by 4
print(Point(Matrix([[1,-2,0],[4,1,0],[0,0,1]])*Matrix(P)))

# g) Rotation about origin by 30°
print(Point(P.x, P.y).rotate(pi/6))
```

Q.5) Apply each of the following transformations on the point P [3, −1].

```python
# Method 1
P=Point(3,-1)

# a) Reflection X-axis
print(Point(P.x,-P.y))

# b) Scale X by 2
print(Point(2*P.x,P.y))

# c) Scale Y by 1.5
print(Point(P.x,1.5*P.y))

# d) Shear Y by 2
print(Point(P.x,P.y+2*P.x))

# e) Scale X by 1/2, Y by 3
print(Point(P.x/2,3*P.y))

# f) Shear X=-2, Y=4
print(Point(P.x-2*P.y,P.y+4*P.x))

# g) Rotation 30°
theta=pi/6
print(Point(P.x*cos(theta)-P.y*sin(theta),P.x*sin(theta)+P.y*cos(theta)))
```

```python
# Method 2
P = Point(3,-1)

# a) Reflection through X-axis
print(P.reflect(Line(Point(0,0),Point(1,0))))

# b) Scaling in X by 2
print(Point(Matrix([[2,0],[0,1]])*Matrix(P)))

# c) Scaling in Y by 1.5
print(Point(Matrix([[1,0],[0,1.5]])*Matrix(P)))

# d) Shearing in Y by 2
print(Point(Matrix([[1,0],[2,1]])*Matrix(P)))

# e) Scaling X by 1/2 and Y by 3
print(Point(Matrix([[1/2,0],[0,3]])*Matrix(P)))

# f) Shearing X by −2 and Y by 4
print(Point(Matrix([[1,-2],[4,1]])*Matrix(P)))

# g) Rotation about origin by 30°
print(P.rotate(pi/6))
```
