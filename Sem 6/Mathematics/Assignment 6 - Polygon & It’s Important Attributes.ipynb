```python
from sympy import *
from sympy.geometry import *
x, y = symbols('x y')
```

Q1) Draw a polygon with vertices (0,0), (1,0), (2,2), (1,4) and find its area & perimeter.

```python
P = Polygon(Point(0,0), Point(1,0), Point(2,2), Point(1,4))
print("Area:", P.area)
print("Perimeter:", P.perimeter)
```

Q2) Draw a regular polygon with 8 sides & radius 5 centred at origin and find its area & perimeter.

```python
P = RegularPolygon(Point(0,0), 5, 8)
print("Area:", P.area)
print("Perimeter:", P.perimeter)
```

Q3) Draw a regular polygon with 7 sides and radius 1.5 centered at (2,2) & reflect it through line x − y = 5.

```python
P = RegularPolygon(Point(2,2), 1.5, 7)
print(P.reflect(Line(x - y - 5)))
```

Q4) Draw a polygon with vertices (0,0), (2,0), (2,3), (1,6) and rotate by 180 degrees.

```python
P = Polygon(Point(0,0), Point(2,0), Point(2,3), Point(1,6))
print(P.rotate(pi))
```

Q5) Reflect the polygon ABC through the line y = 3, where A[1,0], B[2,-1], C[-1,3].

```python
T = Polygon(Point(1,0), Point(2,-1), Point(-1,3))
print(T.reflect(Line(y - 3)))
```

Q6) Find the angle at each vertices of the triangle ABC, where A[0,0], B[2,2], C[0,2].

```python
T = Triangle(Point(0,0), Point(2,2), Point(0,2))
print(T.angles)
```

Q7 Check the following polygons (triangles) are Isosceles, Right angled and Scalene triangle or not

```python
# (i) A[0,0], B[4,0], C[2,4]

T = Triangle(Point(0,0), Point(4,0), Point(2,4))
print(T.is_isosceles(), T.is_right(), T.is_scalene())
```

```python
# (ii) A[0,0], B[4,0], C[4,3]

T = Triangle(Point(0,0), Point(4,0), Point(4,3))
print(T.is_isosceles(), T.is_right(), T.is_scalene())
```

```python
# (iii) A[0,0], B[4,0], C[1,4]
T = Triangle(Point(0,0), Point(4,0), Point(1,4))
print(T.is_isosceles(), T.is_right(), T.is_scalene())
```
