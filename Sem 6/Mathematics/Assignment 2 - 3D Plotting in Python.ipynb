# Assignment 2: 3D Plotting in Python

#### Q1.1 - 3D Contour: f(x,y) = x³ + y²,  x,y ∈ [0, 2π]

```python
import numpy as np
import matplotlib.pyplot as plt

x = y = np.linspace(-2*np.pi, 2*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = X**3 + Y**2

ax = plt.figure(figsize=(8, 6)).add_subplot(111, projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='magma')
ax.set(xlabel='X axis', ylabel='Y axis', zlabel='Z axis', title='f(x,y) = x³ + y²')
plt.show()
```

#### Q1.7 - 3D Contour: f(x,y) = e^(-x²-y²),  x,y ∈ (-1, 1)

#### Q1.8 - 3D Contour: f(x,y) = (sin(4x) - cos(5y)) / 5,  x,y ∈ (-1, 1)

```python
import numpy as np
import matplotlib.pyplot as plt

x = y = np.linspace(-1, 1, 100)
X, Y = np.meshgrid(x, y)
Z = (np.sin(4*X) - np.cos(5*Y)) / 5

ax = plt.figure(figsize=(8, 6)).add_subplot(111, projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='viridis')
ax.set(xlabel='X axis', ylabel='Y axis', zlabel='Z axis', title='f(x,y) = (sin(4x) - cos(5y)) / 5')
plt.show()
```
