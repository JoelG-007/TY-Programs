Q1. Generate 3D contour plots for the following functions:

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# f(x, y) = x^3 + y^2

x = y = np.linspace(0, 2*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = X**3 + Y**2
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```

```python
# f(x, y) = sin(x^2 + y)

x = y = np.linspace(-2*np.pi, 2*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X**2 + Y)
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```

```python
# f(x, y) = sin(x + y) + cos(x)

x = y = np.linspace(-2*np.pi, 2*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X + Y) + np.cos(X)
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```

```python
# f(x, y) = sin(x) + cos(x)

x = y = np.linspace(-2*np.pi, 2*np.pi, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) + np.cos(X)
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```

```python
# f(x, y) = sin(x^2 + y^2)

x = y = np.linspace(-6, 6, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X**2 + Y**2)
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```

```python
# f(x, y) = x^2 + y^2

x = y = np.linspace(-6, 6, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```

```python
# f(x, y) = e^(-x^2 - y^2)

x = y = np.linspace(-1, 1, 100)
X, Y = np.meshgrid(x, y)
Z = np.exp(-X**2 - Y**2)
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```

```python
# f(x, y) = (sin(4x) - cos(5y)) / 5

x = y = np.linspace(-1, 1, 100)
X, Y = np.meshgrid(x, y)
Z = (np.sin(4*X) - np.cos(5*Y)) / 5
plt.contourf(X, Y, Z)
plt.colorbar()
plt.show()
```
