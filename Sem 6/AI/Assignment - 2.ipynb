#### Q1

```python
# Knowledge Representation - Production Rules
# Smart Home System
# Facts:
#   time: evening | presence: home
#   temperature: 17°C | clock: 21
# Rules:
#   Rule 1: evening + home  -> Turn on the lights
#   Rule 2: temp < 18       -> Turn on the heating
#   Rule 3: clock >= 21     -> Lower the thermostat

facts = {
    "time":        "evening",
    "presence":    "home",
    "temperature": 17,
    "clock":       21
}

def rule1(f): return "Turn on the lights"   if f["time"] == "evening" and f["presence"] == "home" else None
def rule2(f): return "Turn on the heating"  if f["temperature"] < 18  else None
def rule3(f): return "Lower the thermostat" if f["clock"] >= 21        else None

def inference_engine(facts):
    actions = []
    for rule in [rule1, rule2, rule3]:
        action = rule(facts)
        if action:
            actions.append(action)
    return actions

print("Triggered Actions")
for act in inference_engine(facts):
    print("-", act)
```

```python
# Knowledge Representation - Production Rules
# Medical Diagnosis System
# Facts:
#   Fever, Cough, Headache, Chest Pain -> all True
# Rules:
#   Rule 1: Fever + Cough + Headache        -> Flu
#   Rule 2: Fever + Headache + Chest Pain   -> Pneumonia
#   Rule 3: Headache + Cough                -> Common Cold

facts = {
    "Fever":      True,
    "Cough":      True,
    "Headache":   True,
    "Chest Pain": True
}

def rule_flu(f):         return "Diagnosis: Flu"         if f["Fever"] and f["Cough"] and f["Headache"]              else None
def rule_pneumonia(f):   return "Diagnosis: Pneumonia"   if f["Fever"] and f["Headache"] and f["Chest Pain"]         else None
def rule_common_cold(f): return "Diagnosis: Common Cold" if f["Headache"] and f["Cough"]                             else None

def inference_engine(facts):
    actions = []
    for rule in [rule_flu, rule_pneumonia, rule_common_cold]:
        action = rule(facts)
        if action:
            actions.append(action)
    return actions

print("Triggered Actions")
for act in inference_engine(facts):
    print("-", act)
```
