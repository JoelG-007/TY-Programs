```python
# Tic-Tac-Toe Game (Two Player)
# 3x3 board, turn-based, checks win/draw

import random

board = [" "] * 9
player = random.choice(["X", "O"])
print(f"Player {player} starts the game")

WIN_COMBOS = [
    [0,1,2],[3,4,5],[6,7,8],  # rows
    [0,3,6],[1,4,7],[2,5,8],  # columns
    [0,4,8],[2,4,6]           # diagonals
]

def display():
    print()
    for i in range(0, 9, 3):
        print(f" {board[i]} | {board[i+1]} | {board[i+2]} ")
        if i < 6:
            print("---|---|---")
    print()

def check_win():
    return any(board[a] == board[b] == board[c] != " " for a, b, c in WIN_COMBOS)

while True:
    display()
    pos = int(input(f"Player {player}, enter position (1-9): ")) - 1

    if pos < 0 or pos > 8 or board[pos] != " ":
        print("Invalid move. Try again.")
        continue

    board[pos] = player

    if check_win():
        display()
        print(f"Player {player} wins!")
        break

    if " " not in board:
        display()
        print("It's a draw!")
        break

    player = "O" if player == "X" else "X"
```

```python
# 8 Puzzle Game (3x3 sliding tiles)
# Initial State: 1 2 3 / 4 _ 5 / 6 7 8
# Goal State:    1 2 3 / 4 5 6 / 7 8 _

board = [1, 2, 3, 4, 0, 5, 6, 7, 8]
goal  = [1, 2, 3, 4, 5, 6, 7, 8, 0]

def show():
    print()
    for i in range(0, 9, 3):
        print(' '.join('_' if board[i+j] == 0 else str(board[i+j]) for j in range(3)))
    print()

show()

while True:
    if board == goal:
        print("Puzzle Solved!")
        break

    try:
        tile = int(input("Enter tile number to move: "))
    except ValueError:
        print("Enter a valid number.")
        continue

    if tile not in board:
        print("Invalid tile.")
        continue

    tile_pos = board.index(tile)
    zero_pos = board.index(0)

    row_t, col_t = divmod(tile_pos, 3)
    row_z, col_z = divmod(zero_pos, 3)

    # Valid move: tile must be adjacent (same row/col, 1 step away)
    if abs(row_t - row_z) + abs(col_t - col_z) == 1:
        board[zero_pos], board[tile_pos] = board[tile_pos], board[zero_pos]
        show()
    else:
        print("Move not allowed.")
```
