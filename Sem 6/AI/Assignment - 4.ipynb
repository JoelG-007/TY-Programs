### Q1

```python
# Simple Chatbot - Student FAQ Automation
# AI Concept: Natural Language Processing (rule-based)

responses = {
    "exam":       "Exam registration is closed.",
    "fee":        "Fees can be paid online through the college portal.",
    "attendance": "Minimum 75% attendance is required.",
    "library":    "Library is open from 9 AM to 6 PM."
}

def chatbot(user_input):
    query = user_input.lower()
    for keyword, reply in responses.items():
        if keyword in query:
            return reply
    return "Sorry, I don't understand your question."

while True:
    question = input("You: ")
    if question.lower() == "exit":
        print("Chatbot: Thank You!")
        break
    print("Chatbot:", chatbot(question))
```

### Q2

```python
# Self-Driving Car Knowledge Base
# AI Concept: Production Rules / Decision Trees

traffic_light = input("Enter Traffic Light Status (Red/Yellow/Green): ").lower()
obstacle      = input("Enter obstacle (Car/Pedestrian/Cyclist/None): ").lower()
crosswalk     = input("Is crosswalk present? (Yes/No): ").lower()

if traffic_light == "red":
    decision = "Stop"
elif traffic_light == "yellow":
    decision = "Slow Down"
elif crosswalk == "yes" and obstacle == "pedestrian":
    decision = "Stop"
elif obstacle == "pedestrian":
    decision = "Stop"
elif obstacle in ["car", "cyclist"]:
    decision = "Avoid"
elif traffic_light == "green" and obstacle == "none":
    decision = "Go"
else:
    decision = "Proceed with Caution"

print("Decision:", decision)
```

### Q3

```python
# Medical Diagnosis Using Rule-Based Expert System
# AI Concept: Expert System (IF-THEN rules)

symptoms = {
    "fever":      input("Fever? (yes/no): ").lower() == "yes",
    "cough":      input("Cough? (yes/no): ").lower() == "yes",
    "headache":   input("Headache? (yes/no): ").lower() == "yes",
    "chest_pain": input("Chest Pain? (yes/no): ").lower() == "yes"
}

diagnoses = []

if symptoms["fever"] and symptoms["cough"] and symptoms["headache"]:
    diagnoses.append("Flu")
if symptoms["fever"] and symptoms["headache"] and symptoms["chest_pain"]:
    diagnoses.append("Pneumonia")
if symptoms["headache"] and symptoms["cough"]:
    diagnoses.append("Common Cold")

if diagnoses:
    print("\nPossible Diagnoses:")
    for d in diagnoses:
        print("-", d)
else:
    print("\nNo diagnosis found. Please consult a doctor.")
```
