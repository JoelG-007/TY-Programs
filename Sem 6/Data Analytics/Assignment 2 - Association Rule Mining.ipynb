```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
def run_apriori(transactions, support=0.4, confidence=0.6):
    # Encode transactions into one-hot DataFrame
    te = TransactionEncoder()
    df = pd.DataFrame(
        te.fit(transactions).transform(transactions),
        columns=te.columns_
    )
    print("\nEncoded Dataset:\n", df)

    # Find frequent itemsets using Apriori
    freq = apriori(df, min_support=support, use_colnames=True)
    print("\nFrequent Itemsets:\n", freq)

    # Generate association rules
    rules = association_rules(freq, metric="confidence", min_threshold=confidence)
    print("\nAssociation Rules:")
    print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```

```python
def load_transactions(filepath):
    df = pd.read_csv(filepath)
    # Drop empty rows
    df.dropna(how='all', inplace=True)
    # Convert each row to a list of non-null string values
    return [
        [col for col in df.columns if isinstance(df.at[i, col], str)]
        for i in range(len(df))
    ]
```

```python
print("\n--- Market Basket Optimisation ---")
run_apriori(load_transactions("Market_Basket_Optimisation.csv"), support=0.01, confidence=0.5)
```
