```python
# importing necessary libraries
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules    # aprirori for frequent itemset mining and association_rules for generating rules
from mlxtend.preprocessing import TransactionEncoder                # TransactionEncoder for converting the list of transactions into a format suitable for apriori algorithm
```

```python
transactions = [
    ['Bread', 'Milk'],
    ['Bread', 'Diaper', 'Beer', 'Eggs'],
    ['Milk', 'Diaper', 'Beer', 'Coke'],
    ['Bread', 'Milk', 'Diaper', 'Beer'],
    ['Bread', 'Milk', 'Diaper', 'Coke']
]
```

```python
# Converting Categorical Values to Numeric Format
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)  # Transforming the list of transactions into a one-hot encoded format

df = pd.DataFrame(te_data, columns=te.columns_)         # Creating a DataFrame from the transformed data with appropriate column names
print("Encoded Dataset:")
print(df)
```

```python
# Applying the Apriori Algorithm (with min_support = 0.4)
# min_support is the minimum proportion of transactions that must contain an itemset for it to be considered frequent.
frequent_itemsets = apriori(
    df,
    min_support=0.4,            # min_support is set to 0.4, meaning we are looking for itemsets that appear in at least 40% of the transactions
    use_colnames=True
)

print("\nFrequent Itemsets (min_support = 0.4):")
print(frequent_itemsets)
```

```python
# Generating Association Rules (with min_threshold = 0.6 for confidence)
rules = association_rules(
    frequent_itemsets,          # frequent_itemsets is the result from the apriori algorithm that contains the frequent itemsets and their support values
    metric="confidence",        # metric is set to "confidence" to evaluate the strength of the rules based on confidence
    min_threshold=0.6           # min_threshold is set to 0.6, meaning we are looking for rules that have a confidence of at least 60%
)

print("\nAssociation Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
'''
With reference to the output of the association rules, the following columns are important for understanding the rules generated:
antecedents: The item(s) on the left-hand side of the rule (the "if" part).
consequents: The item(s) on the right-hand side of the rule (the "then" part).
support: The proportion of transactions that contain the itemset (both antecedents and consequents).
confidence: The likelihood that the consequents are purchased when the antecedents are purchased (calculated as support of the rule divided by support of the antecedents).
lift: The ratio of the observed support to that expected if the antecedents and consequents were independent (calculated as confidence divided by the support of the consequents).
'''
```

```python
# Repeating Apriori with a Different Minimum Support (min_support = 0.6)
# Why? - To see how changing the minimum support threshold affects the frequent itemsets generated. \
# A higher minimum support will result in fewer itemsets being considered frequent, which can help identify stronger associations 
# but may also miss less common but potentially interesting itemsets.
frequent_itemsets_06 = apriori(
    df,
    min_support=0.6,
    use_colnames=True
)

print("\nFrequent Itemsets (min_support = 0.6):")
print(frequent_itemsets_06)
```

```python
# (min_support = 0.8)
# Why? - To further increase the minimum support threshold to see how it affects the frequent itemsets.
# A very high minimum support may result in very few or no itemsets being considered frequent,
# which can help identify only the most common associations but may miss many interesting itemsets that are less common.
frequent_itemsets_08 = apriori(
    df,
    min_support=0.8,
    use_colnames=True
)

print("\nFrequent Itemsets (min_support = 0.8):")
print(frequent_itemsets_08)
```

```python
'''
Trying for another dataset with different transactions to see how the results change with different data and 
to compare the results with the original dataset.

This will help in understanding the impact of different transaction patterns on the frequent itemsets and association rules generated.
'''
```

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# 1. Market Basket Analysis
# Load Market Basket dataset
market_df = pd.read_csv("market_basket.csv")

print("Dataset Info:")
print(market_df.info())

print("\nFirst 5 rows:")
print(market_df.head())
```

```python
# Drop null values
market_df.dropna(inplace=True)

# Convert rows to list of transactions
transactions = market_df.iloc[:, 0].apply(lambda x: x.split(',')).tolist()
```

```python
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)

encoded_df = pd.DataFrame(te_data, columns=te.columns_)
print(encoded_df.head())
```

```python
frequent_itemsets = apriori(
    encoded_df,
    min_support=0.02,
    use_colnames=True
)

print("Frequent Itemsets:")
print(frequent_itemsets)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.5
)

print("Association Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
# Groceries Dataset
# Load Groceries dataset
groceries_df = pd.read_csv("groceries.csv")

print("Dataset Info:")
print(groceries_df.info())

print("\nFirst 5 rows:")
print(groceries_df.head())
```
